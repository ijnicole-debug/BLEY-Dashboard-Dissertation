## Correlation Analysis

```python
#load needed libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
df = pd.read_csv('../data_clean/master_dataset.csv')
```

```python
#correlation analysis
corr = df.drop(columns=['la_code', 'borough']).corr()
corr
```

```python
#plot correlation matrix
plt.figure(figsize=(10,8))

sns.heatmap (corr, annot=True, cmap='RdYlGn', center=0, fmt='.2f')

plt.title('Correlation Matrix')
plt.show()
```

```python
#city of London analysis
#create flag
df['city_of_london_flag'] = (
    df['borough'] == 'City of London'
).astype(int)
```

```python
#without City of London
df_no_city = df[df['borough'] != 'City of London']
```
