## DATA CLEANING- CHILDCARE PLACES PER 100 UNDER 5s

```python
#import libraries
import pandas as pd

# Load dataset
df = pd.read_csv('../data_clean/ons_under5.csv')
df_places = pd.read_csv('../data_clean/ofsted_places.csv')
```

```python
#quick preview
print(df.head())
```

```python
#quick preview
print(df_places.head())
```

```python
#merge data
df = pd.merge(df, df_places, on='borough', how='inner')
```

```python
#calculate childcare access
df['childcare_per_100'] = (df['total_places'] / df['under5'] * 100)
```

```python
#final structure
df_clean = df[['borough', 'childcare_per_100']]
df_clean = df_clean.sort_values(by='borough').reset_index(drop=True)
```

```python
#validate
print("Rows:", len(df_clean))
print(df_clean.head(33))
```

```python
#save merged dataset
df_clean.to_csv('../data_clean/childcare_access.csv', index=False)
```
