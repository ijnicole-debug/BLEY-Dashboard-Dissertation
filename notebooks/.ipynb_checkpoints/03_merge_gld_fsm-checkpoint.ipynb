## Merge- GLD & FSM

```python
#import libraries
import pandas as pd

#load dataset
gld = pd.read_csv('../data_clean/gld_clean.csv')
fsm = pd.read_csv('../data_clean/fsm_clean.csv')
```

```python
#data validation
gld.head()
```

```python
#data validation
fsm.head()
```

```python
#merge
df_merged = pd.merge(gld, fsm, on='la_code', how='inner')
```

```python
#validate merge
print("Merged rows:", len(df_merged))
```

```python
#validation of borough names
print("Borough match check:",
      (df_merged['borough_x'] == df_merged['borough_y']).all())

print(df_merged.head())
```

```python
#clean duplicate columns
df_merged = df_merged.drop(columns=['borough_y'])
df_merged = df_merged.rename(columns={'borough_x': 'borough'})
```

```python
#final structure check
print(df_merged.head())
```

```python
#save merged data
df_merged.to_csv('../data_clean/gld_fsm_merged.csv', index=False)
```

```python
#Quick analysis
print("Correlation:")
print(df_merged[['gld', 'fsm']].corr())
```

## Merge SEN to master dataset

```python
#load dataset
sen = pd.read_csv('../data_clean/sen_clean.csv')
```

```python
#merge
df_merged = pd.merge(df_merged, sen, on='la_code', how='inner')
```

```python
#validate
print("Rows after SEN merge:", len(df_merged))
df_merged.head()
```

```python
#clean duplicate columns
df_merged = df_merged.drop(columns=['borough_y'])
df_merged = df_merged.rename(columns={'borough_x': 'borough'})
```

```python
#save merged
df_merged.to_csv('../data_clean/gld_fsm_sen_merged.csv', index=False)
```

```python
#validate
df_merged.head()
```

```python
#top 5 boroughs by GLD
df_merged.sort_values(by='gld', ascending=False).head()
```

```python
#lowest 5 boroughs by GLD
df_merged.sort_values(by='gld', ascending=True).head()
```

```python
#Top 5 by FSM
df_merged.sort_values(by='fsm', ascending=False).head()
```

```python
#top 5 by SEN
df_merged.sort_values(by='sen', ascending=False).head()
```

```python
#correlation
df_merged.corr(numeric_only=True)
```
