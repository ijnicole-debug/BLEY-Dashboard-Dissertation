## Data Cleaning- FSM

```python
##Import libraries
import pandas as pd

##load dataset
df = pd.read_csv('../data_raw/spc_fsm6.csv')

#inspect dataset
df.head()
```

```python
#confirm column types
df.dtypes
```

```python
#filter by time and geographic level
df = df[df['geographic_level'] == 'Local authority']
df = df[df['time_period'] == 202223]
```

```python
#remove breakdowns
df = df[df['breakdown_topic'] == 'Total']
df = df[df['breakdown'] == 'Total']
df = df[df['phase_type_grouping'] == 'Total']
```

```python
#keep FSM6 only
df = df[df['fsm6_eligibility'] == 'FSM6']
```

```python
#filter London boroughs
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham',
    'Islington', 'Kensington and Chelsea', 'Lambeth', 'Lewisham',
    'Southwark', 'Tower Hamlets', 'Wandsworth', 'Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent',
    'Bromley', 'Croydon', 'Ealing', 'Enfield', 'Haringey',
    'Harrow', 'Havering', 'Hillingdon', 'Hounslow',
    'Kingston upon Thames', 'Merton', 'Newham', 'Redbridge',
    'Richmond upon Thames', 'Sutton', 'Waltham Forest',
    'City of London'
]
df = df[df['la_name'].isin(london_boroughs)]
```

```python
#select variables
df_clean = df[['new_la_code', 'la_name', 'percent_of_pupils']]
```

```python
#rename
df_clean.columns = ['la_code', 'borough', 'fsm']
```

```python
#clean and sort
df_clean['borough'] = df_clean['borough'].str.strip()
df_clean = df_clean.sort_values(by='borough').reset_index(drop=True)
```

```python
#save clean dataset
df_clean.to_csv('../data_clean/fsm_clean.csv', index=False)
```
