```python
#load master dataset
df = pd.read_csv('../data_clean/master_dataset.csv')
```

## #Bivariate OLS (Simple regression)

```python
#select variables
variables = ['fsm', 'sen', 'idaci', 'ofsted', 'childcare_per_100', 'ethnic_diversity']

# variables where HIGH = BAD (colour logic flips)
negative_vars = ['fsm', 'idaci', 'sen']

#store regression results
results = []
```

```python
# Individual Variable Analysis- Scatter Plots, Simple OLS Regression and Borough Rankings
# Store regression results
results = []
# Loop through variables
for var in variables:

#Create figure with two panels
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

#Define x and y variables
    x = df[var]
    y = df["gld"]

#LEFT PANEL
# Scatter Plot + Regression Line
#Scatter plot
    axes[0].scatter(x, y)

#Fit regression line
    z = np.polyfit(x, y, 1)
    p = np.poly1d(z)

    axes[0].plot(x, p(x), linewidth=2)

#Simple OLS Regression
    X_reg = sm.add_constant(df[[var]])

    model = sm.OLS(y, X_reg).fit()

#Extract regression statistics
    coef = round(model.params[var], 3)
    pval = round(model.pvalues[var], 3)
    r2 = round(model.rsquared, 3)

#Store regression results
    results.append({

        "Variable": var,
        "Coefficient": coef,
        "P_value": pval,
        "R_squared": r2

    })

# Labels
    axes[0].set_xlabel(var.replace("_", " ").title())
    axes[0].set_ylabel("Good Level of Development (GLD)")

    axes[0].set_title(
        f"{var.replace('_',' ').title()} vs GLD\n"
        f"Coefficient = {coef} | p = {pval} | R² = {r2}"
    )
# RIGHT PANEL
# Borough Ranking
    df_sorted = df.sort_values(by=var)

#Define tertiles
    low_threshold = df_sorted[var].quantile(0.33)
    high_threshold = df_sorted[var].quantile(0.66)

    colours = []

    for value in df_sorted[var]:

        if var in negative_vars:

            # Lower values are more favourable

            if value <= low_threshold:
                colours.append("green")

            elif value >= high_threshold:
                colours.append("red")

            else:
                colours.append("orange")

        else:

            # Higher values are more favourable

            if value <= low_threshold:
                colours.append("red")

            elif value >= high_threshold:
                colours.append("green")

            else:
                colours.append("orange")

#Horizontal bar chart
    axes[1].barh(
        df_sorted["borough"],
        df_sorted[var],
        color=colours
    )

    axes[1].set_xlabel(var.replace("_", " ").title())

    axes[1].set_title(
        f"{var.replace('_',' ').title()} across London Boroughs\n"
        "(Green = favourable | Amber = moderate | Red = less favourable)"
    )

    plt.tight_layout()

    plt.savefig(
        f"../outputs/{var}_visualisation.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```

```python
# REGRESSION SUMMARY TABLE
results_df = pd.DataFrame(results)
results_df
```

## Model Diagnostics and Sensitivity Analysis

The full model is refitted here into its own variables, because `model`, `X` and `y`
are reused by the individual variable analysis above.

```python
#additional libraries for diagnostics
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import het_breuschpagan
from scipy import stats
```

```python
#refit the full model into its own variables
predictors = ['fsm', 'sen', 'idaci', 'ofsted', 'childcare_per_100', 'ethnic_diversity']

y_full = df['gld']
X_full = sm.add_constant(df[predictors])
model_full = sm.OLS(y_full, X_full).fit()

print(model_full.summary())
```

```python
#multicollinearity check
vif = pd.DataFrame({
    'variable': predictors,
    'VIF': [variance_inflation_factor(X_full.values, i) for i in range(1, X_full.shape[1])]
}).round(2)

vif
```

```python
#standardised coefficients, so variables on different scales can be compared
sd_y = y_full.std(ddof=1)

standardised = pd.DataFrame({
    'variable': predictors,
    'coefficient': [model_full.params[v] for v in predictors],
    'std_error': [model_full.bse[v] for v in predictors],
    'p_value': [model_full.pvalues[v] for v in predictors],
    'standardised_beta': [model_full.params[v] * df[v].std(ddof=1) / sd_y for v in predictors]
}).round(4)

standardised
```

```python
#assumption checks on the full model
shapiro_stat, shapiro_p = stats.shapiro(model_full.resid)
bp = het_breuschpagan(model_full.resid, model_full.model.exog)

print('Normality of residuals (Shapiro-Wilk): W = %.3f, p = %.3f' % (shapiro_stat, shapiro_p))
print('Constant variance (Breusch-Pagan):     p = %.3f' % bp[1])
print('Highest variance inflation factor:     %.2f' % vif['VIF'].max())
print()
print('Values above 0.05 for the first two indicate the assumption is not violated.')
```

### Reduced model

With 32 boroughs, six predictors leave roughly five observations each, which is below
the level normally considered adequate. A reduced specification is fitted as a
sensitivity analysis.

```python
#reduced model: childcare accessibility and FSM only
X_reduced = sm.add_constant(df[['childcare_per_100', 'fsm']])
model_reduced = sm.OLS(y_full, X_reduced).fit()

comparison = pd.DataFrame({
    'model': ['Full (6 predictors)', 'Reduced (2 predictors)'],
    'r_squared': [model_full.rsquared, model_reduced.rsquared],
    'adj_r_squared': [model_full.rsquared_adj, model_reduced.rsquared_adj],
    'f_statistic': [model_full.fvalue, model_reduced.fvalue],
    'p_value': [model_full.f_pvalue, model_reduced.f_pvalue]
}).round(4)

comparison
```

### Leave-one-out sensitivity check

Each significant bivariate correlation is recalculated 32 times, removing one borough
at a time, to establish whether any single borough is driving the result.

```python
#leave-one-out check on the significant correlations
sensitivity = []

for var in ['childcare_per_100', 'idaci', 'fsm']:
    base_r, base_p = stats.pearsonr(df[var], df['gld'])
    worst_borough, worst_r, worst_p, largest_shift = None, base_r, 1.0, 0.0

    for i in df.index:
        subset = df.drop(index=i)
        r, p = stats.pearsonr(subset[var], subset['gld'])
        if abs(base_r - r) > largest_shift:
            worst_borough, worst_r, worst_p = df.loc[i, 'borough'], r, p
            largest_shift = abs(base_r - r)

    sensitivity.append({
        'variable': var,
        'r_all_boroughs': base_r,
        'p_all_boroughs': base_p,
        'most_influential': worst_borough,
        'r_without': worst_r,
        'p_without': worst_p,
        'still_significant': worst_p < 0.05
    })

sensitivity_df = pd.DataFrame(sensitivity).round(4)
sensitivity_df
```
