## Data Cleaning- House prices

```python
#import library
import pandas as pd

#load dataset
df = pd.read_csv('../data_raw/Average-prices-2026-04.csv')

#quick preview
df.head()
```

```python
#confirm column types
df.dtypes
```

```python
#confirm completeness
df.info()
```

```python
#convert date
df["Date"] = pd.to_datetime(df["Date"])
```

```python
#select the required year
df = df[(df["Date"] >= "2025-01-01") &(df["Date"] <= "2025-12-31")]
```

```python
#filter london broroughs
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham',
    'Islington', 'Kensington and Chelsea', 'Lambeth', 'Lewisham',
    'Southwark', 'Tower Hamlets', 'Wandsworth', 'City of Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent',
    'Bromley', 'Croydon', 'Ealing', 'Enfield', 'Haringey',
    'Harrow', 'Havering', 'Hillingdon', 'Hounslow',
    'Kingston upon Thames', 'Merton', 'Newham', 'Redbridge',
    'Richmond upon Thames', 'Sutton', 'Waltham Forest'
]

df = df[df['Region_Name'].isin(london_boroughs)]
```

```python
#select columns needed
df = df[['Region_Name', 'Area_Code', 'Average_Price']]
```

```python
#reconfirm completeness
df.info()
```

```python
#rename columns
df = df.rename(columns={
    "Region_Name": "borough",
    "Area_Code": "la_code",
    "Average_Price": "average_house_price"
})
```

```python
#calculate average price for each borough
df = (
    df.groupby(
        ["borough", "la_code"],
        as_index=False
    )["average_house_price"]
    .mean()
)
```

```python
#round prices to whole pounds
df["average_house_price"] = (df["average_house_price"].round(0).astype(int))
```
