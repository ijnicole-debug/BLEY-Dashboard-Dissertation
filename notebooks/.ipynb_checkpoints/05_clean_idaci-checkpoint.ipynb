## Data Cleaning- IDACI

```python
#import libraries
import pandas as pd

# Load dataset
df = pd.read_excel('../data_raw/File_5_-_IoD2019_Scores.xlsx', sheet_name='IoD2019 Scores')
```

```python
#preview
df.head()
```

```python
#confirm column types
df.dtypes
```

```python
#select columns needed
df = df[[
    'Local Authority District code (2019)',
    'Local Authority District name (2019)',
    'Income Deprivation Affecting Children Index (IDACI) Score (rate)'
]]
```

```python
#aggregate to borough level
df_la = df.groupby(
    ['Local Authority District code (2019)', 
     'Local Authority District name (2019)'],
    as_index=False
)['Income Deprivation Affecting Children Index (IDACI) Score (rate)'].mean()
```

```python
#rename columns for consistency
df_la.columns = ['la_code', 'borough', 'idaci']
```

```python
#filter London
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham',
    'Islington', 'Kensington and Chelsea', 'Lambeth', 'Lewisham',
    'Southwark', 'Tower Hamlets', 'Wandsworth', 'Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent',
    'Bromley', 'Croydon', 'Ealing', 'Enfield', 'Haringey',
    'Harrow', 'Havering', 'Hillingdon', 'Hounslow',
    'Kingston upon Thames', 'Merton', 'Newham', 'Redbridge',
    'Richmond upon Thames', 'Sutton', 'Waltham Forest'
]
df_la = df_la[df_la['borough'].isin(london_boroughs)]
```

```python
#clean and sort
df_la['borough'] = df_la['borough'].str.strip()
df_la = df_la.sort_values(by='borough').reset_index(drop=True)
```

```python
#validate
print("Rows:", len(df_la))
print(df_la.head())
```

```python
#save
df_la.to_csv('../data_clean/idaci_clean.csv', index=False)
```
