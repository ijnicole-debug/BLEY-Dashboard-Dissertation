## DATA CLEANING- SEN

```python
#import library
import pandas as pd

#load dataset
df = pd.read_csv('../data_raw/sen_phase_type_.csv')

#preview dataset
df.head()
```

```python
#confirm column types
df.dtypes
```

```python
#filter core conditions
df = df[df['geographic_level'] == 'Local authority']
df = df[df['time_period'] == 202223]
```

```python
#remove breakdowns
df = df[df['phase_type_grouping'] == 'Total']
df = df[df['type_of_establishment'] == 'Total']
df = df[df['hospital_school'] == 'Total']
```

```python
#filter london
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham',
    'Islington', 'Kensington and Chelsea', 'Lambeth', 'Lewisham',
    'Southwark', 'Tower Hamlets', 'Wandsworth', 'Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent',
    'Bromley', 'Croydon', 'Ealing', 'Enfield', 'Haringey',
    'Harrow', 'Havering', 'Hillingdon', 'Hounslow',
    'Kingston upon Thames', 'Merton', 'Newham', 'Redbridge',
    'Richmond upon Thames', 'Sutton', 'Waltham Forest',
    'City of London'
]
df = df[df['la_name'].isin(london_boroughs)]
```

```python
# Convert to numeric 
df['sen_support_percent'] = pd.to_numeric(df['sen_support_percent'], errors='coerce')
df['ehc_plan_percent'] = pd.to_numeric(df['ehc_plan_percent'], errors='coerce')
```

```python
#create total SEN
df['sen_total'] = df['sen_support_percent'] + df['ehc_plan_percent']
```

```python
#Select variables
df_clean = df[['new_la_code', 'la_name', 'sen_total']]
```

```python
#rename columns
df_clean.columns = ['la_code', 'borough', 'sen']
```

```python
#clean and sort
df_clean['borough'] = df_clean['borough'].str.strip()
df_clean = df_clean.sort_values(by='borough').reset_index(drop=True)
```

```python
#save clean dataset
df_clean.to_csv('../data_clean/sen_clean.csv', index=False)
```
