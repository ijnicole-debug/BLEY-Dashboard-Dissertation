## Data Cleaning- GLD

```python
#import library
import pandas as pd

#load dataset
df = pd.read_csv('../data_raw/1_eyfsp_headline_measures_2022_2023.csv')

#quick preview
df.head()
```

```python
#confirm column types
df.dtypes
```

```python
#filter to borough level
df = df[df['geographic_level'] == 'Local authority']
```

```python
#filter year 
df = df[df['time_period'] == 202223]
```

```python
#Remove breakdowns
df = df[df['characteristic'] == 'Total']
df = df[df['characteristic_type'] == 'Total']
df = df[df['gender'] == 'Total']
```

```python
#create London borough list
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham',
    'Islington', 'Kensington and Chelsea', 'Lambeth', 'Lewisham',
    'Southwark', 'Tower Hamlets', 'Wandsworth', 'Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent',
    'Bromley', 'Croydon', 'Ealing', 'Enfield', 'Haringey',
    'Harrow', 'Havering', 'Hillingdon', 'Hounslow',
    'Kingston upon Thames', 'Merton', 'Newham', 'Redbridge',
    'Richmond upon Thames', 'Sutton', 'Waltham Forest',
    'City of London'
]
```

```python
#Filter to London only
df = df[df['la_name'].isin(london_boroughs)]
```

```python
#select variables
df_clean = df[['new_la_code', 'la_name', 'gld_percentage']]
```

```python
#rename variables
df_clean.columns = ['la_code', 'borough', 'gld']
```

```python
#clean names
df_clean['borough'] = df_clean['borough'].str.strip()
```

```python
#save clean data
df_clean.to_csv('../data_clean/gld_clean.csv', index=False)
```
